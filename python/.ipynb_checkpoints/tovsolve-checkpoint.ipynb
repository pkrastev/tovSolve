# tovSolve in Python: enthalpy formalism

This notebook uses [`tovsolve.py`](tovsolve.py), the Python version of the enthalpy-formalism solver
(`tov_h.x` / `c/tov_h_c.x`), to compute neutron-star sequences for all EOS tables in the repository and to
draw the figures shown in the [README](../README.md).

For each central baryon density the solver integrates the TOV equations with the pseudo-enthalpy *h* as the
independent variable (Lindblom 1992), together with the equations for the tidal Love number *k*<sub>2</sub> and
the moment of inertia *I*. The Python code reproduces the Fortran and C results **bit for bit**. It is compiled
with Numba when available (close to C speed) and otherwise runs as plain Python.

**Requirements:** numpy, matplotlib; numba (optional, recommended).

```python
import sys, time
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
from IPython.display import Markdown, display

%matplotlib inline

ROOT = Path.cwd().parent if Path.cwd().name == "python" else Path.cwd()
sys.path[:0] = [str(ROOT / "python"), str(ROOT / "scripts")]

import tovsolve
import make_figures as mf   # the plotting code behind the README figures

print("repository:", ROOT.name)
print("Numba acceleration:", tovsolve.NUMBA)
```

## One star

`solve_tov_h(eos, n_c)` returns a `Star` for a central baryon density *n*<sub>c</sub> in fm<sup>−3</sup>:
mass (M<sub>☉</sub>), radius (km), *k*<sub>2</sub>, *λ* (10<sup>36</sup> g cm<sup>2</sup> s<sup>2</sup>),
*I* (10<sup>45</sup> g cm<sup>2</sup>), compactness *β* and *n*<sub>c</sub>. The dimensionless tidal
deformability is Λ = (2/3) *k*<sub>2</sub> *β*<sup>−5</sup>.

```python
eos = tovsolve.Eos(ROOT / "eos_SLY4.in")
print(eos)

star = tovsolve.solve_tov_h(eos, 0.5)
print(star)
print(f"Lambda = {2/3 * star.k2 / star.beta**5:.1f}")
```

## A sequence of stars

`sequence(eos, rho_start, rho_end, nsteps)` uses the same density steps as the Fortran and C drivers and
returns a dict of numpy arrays. `format_header()` and `format_row()` print the familiar fixed-width table
(identical to the output of `tov_h.x`).

```python
t0 = time.perf_counter()
seq = tovsolve.sequence(eos, 0.3, 1.2, 10)
print(f"{len(seq['mass'])} stars in {time.perf_counter() - t0:.3f} s\n")

print(tovsolve.format_header())
for s in zip(*(seq[k] for k in tovsolve.Star._fields)):
    print(tovsolve.format_row(s))
```

## All EOS tables

281 central densities from 0.10 to 1.50 fm<sup>−3</sup> for each of the eight tables. `mf.load_all` keeps the
stable branch (up to the maximum mass) and drops configurations lighter than 0.2 M<sub>☉</sub>.

```python
t0 = time.perf_counter()
data = mf.load_all("python")
print(f"{len(data)} EOS tables, {len(data) * mf.NSTEPS} stars in {time.perf_counter() - t0:.1f} s")

display(Markdown(mf.summary_table(data)))
```

## Figures

The same figures as in the README. Each has two panels, the MDI family (blue ramp ordered by *x*) and the
four other EOSs, with the other group drawn as gray context. The stable branch ends with a dot at
*M*<sub>max</sub>. Observational constraints: PSR J0740+6620, *M* = 2.08 ± 0.07 M<sub>☉</sub>
(Fonseca et al. 2021), and GW170817, Λ<sub>1.4</sub> = 190<sup>+390</sup><sub>−120</sub>
(Abbott et al. 2018). Set `MODE = "dark"` for the dark versions.

```python
MODE = "light"   # or "dark"

for spec in mf.FIGS:
    fig = mf.make_figure(data, spec, MODE, solver="python", save=False)
    fig.set_dpi(80)
    display(fig)
    plt.close(fig)
```

To write the PNG files used by the README (light and dark versions, in `figures/`), set `SAVE = True`:

```python
SAVE = False

if SAVE:
    for spec in mf.FIGS:
        for mode in ("light", "dark"):
            path = mf.make_figure(data, spec, mode, solver="python")
            print("wrote", Path(path).relative_to(ROOT))
```

## Check against the C solver

If the C version is built (`make -C c`), compare the printed output of the two solvers for every star of every EOS. (At full double
precision they are bit-identical too; see the README.)

```python
import subprocess

exe = ROOT / "c" / "tov_h_c.x"
if not exe.exists():
    print("c/tov_h_c.x not found: build it with `make -C c` to run this check")
else:
    for key, fname, label, *_ in mf.EOS:
        py = mf.rows_python(fname)
        c = mf.rows_c(fname)
        print(f"{label:15s} printed output identical: {py == c}")
```
